```python
"""
Connects and disconnects to/from neaSNOM if executed in IPython/Jupyter console.
To control neaSNOM from console, remove `nea_tools.disconnect()`
Requires full SDK to be enabled
"""

import asyncio
import nest_asyncio
import nea_tools

# provide empty string to look for dll files in default locations on server or in current working directory
# or provide an absolute path like r"C:\ClientApplication_1_10_8738_0"  
path_to_dll = None
fingerprint = None
host = 'nea-server'

# connecting and creating module neaspec on success
# inserting async task in IPython event loop
loop = asyncio.get_event_loop()
nest_asyncio.apply(loop)
loop.run_until_complete(nea_tools.connect(host,fingerprint,path_to_dll))
```
